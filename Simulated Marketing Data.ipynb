```python
import faker as f
import pandas as pd
import numpy as np
import seaborn as sns
from random import seed

fake=f.Faker()

seed(40)
# Generating names column
names = []
for i in range(0,101):
    names.append(fake.name())

# Generating emails column
emails = [(lambda n: n.replace(" ","")+"@fakemail.com")(name) for name in names].copy()


# Creating fake dictionary
datadict = {"names": pd.Series(names),
            "emails": pd.Series(emails)}


# Putting together dataframe
marketResearch = pd.DataFrame(datadict)
marketResearch
```
